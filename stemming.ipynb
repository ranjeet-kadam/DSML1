### Stemming

- Stemming is the process of reducing the word to its word stem that affixes to suffixes and prefixes or the root of words known as a lemma.

```python
## Classification Problem
## Comments on the product is positive or negative
## Review --> eating --> eat, eaten 
words = ["eating","eats","eaten","writing","writes","programming","program","History","studies","fried"]
```

### Stemmer Technique
- Porter Stemmer

```python
from nltk import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+ "---->" + stemming.stem(word))
```

##### 2. Regex Stemmer

```python
from nltk import RegexpStemmer
```

```python
regex_Stemmer = RegexpStemmer(regexp="ing$|s$|es$|able$|ion$",min=4)
```

```python
regex_Stemmer.stem('eating')
regex_Stemmer.stem('Projection')
regex_Stemmer.stem('studies')
```

##### 3. Snowball Stemmer

```python
from nltk import SnowballStemmer
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"---->"+snowball_stemmer.stem(word))
```
