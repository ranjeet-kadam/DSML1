```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### Sentences
sent =[
    'I love my dog',
    'I love my cat',
    'the dog is my best friend',
    'I am a good developer',
    'understanding embedding layer',
    'you can use embedding layer in deep learning'
]
```

```python
sent
```

```python
### Defining Vocabulary Size
voc_size = 10000
```

```python
### One hot Representation3
one_hot_repr = [one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
### Word Embedding Representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences 
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
set_length = 8
embedded_docs = pad_sequences(one_hot_repr,padding='pre',maxlen=set_length)
print(embedded_docs)
```

```python
### Feature representation
dim = 10
model = Sequential()
# Specify input_length so the model can be built and summary shows shapes
model.add(Embedding(input_dim=voc_size, output_dim=dim, input_length=set_length))
model.compile(optimizer='adam', loss='mse')
# Explicitly build the model to populate the summary
model.build(input_shape=(None, set_length))
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
