```python
### Data Preprocessing and Cleaning
import re
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
corpus = []
for i in range(0, len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.lower()
    review = review.split()
    review =[ps.stem(word)for word in review if not word in stopwords.words('english')]
    review =' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from nltk.stem import WordNetLemmatizer
lenmmatizer = WordNetLemmatizer()
```

```python
corpus1 = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.lower()
    review = review.split()
    review =[
        lenmmatizer.lemmatize(word) 
        for word in review 
        if word not in stopwords.words('english')
    ]
    review =' '.join(review)
    corpus1.append(review)
```

```python
corpus1
```

```python
### Create a Bag of Word model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=40,binary=True)
```

```python
X=cv.fit_transform(corpus1).toarray()
```

```python
### Create the Bag of Words with N-Grams
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=100,binary=True,ngram_range=(2,3))
X = cv.fit_transform(corpus1).toarray()
```
