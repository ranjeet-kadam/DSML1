```python
sentences = "The Eiffle Tower was built in 1889. It is located in Paris, France. The tower is 324 meters tall. The Gustave Eiffle was the engineer behind the construction of the tower."
```

```python
import nltk

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
words = nltk.word_tokenize(sentences)
```

```python
tag_element = nltk.pos_tag(words)
```

```python
nltk.ne_chunk(tag_element).draw()
```
