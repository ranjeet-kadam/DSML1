```python
import pandas as pd
messages = pd.read_csv(r"D:\Krish_Naik\Python\NLP For ML\uci_sms_spam_mock.csv",sep=',')
```

```python
messages
```

```python
import re 
import nltk
nltk.download('stopwords')
```

```python
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()
```

```python
corpus = []
for i in range(0,len(messages)):
    review = re.sub('[^a-zA-Z]',' ',messages['message'][i])
    review = review.split()
    review = [
        lemma.lemmatize(word) 
        for word in review
        ]
    review = ' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess ### Convert a document into a list of lowercase tokens, ignoring tokens that are too short or too long
```

```python
words =[]
for sent in corpus:
    sent_token =sent_tokenize(sent)
    for sent in sent_token:
        words.append(simple_preprocess(sent))
```

```python
import gensim
```

```python
model=gensim.models.Word2Vec(words)
```

```python
import numpy as np
```

```python
def avg_word2vec(doc):
    # remove out-of-vocabulary words
    #sent =[word for word in doc if word in model.wv.index_to_key]
    #print(sent)
    return np.mean([model.wv[word] for word in doc if word in model.wv.index_to_key],axis=0)
    
```

```python
from tqdm import tqdm
```

```python
# Apply for the entire sentences
X=[]
for i in tqdm(range(len(words))):
    X.append(avg_word2vec(words[i]))
```
