```python
#### 1. Preprocessing and cleaning
## Postive revies is 1 and negative review is 0
data['rating'] =data['rating'].apply(lambda x:0 if x<3 else 1)
```

```python
data['rating'].unique()
```

```python
data['rating'].value_counts()
```

```python
### Step 1 -- Lower all the cases
data['reviewText'] =data['reviewText'].str.lower()
```

```python
data.head()
```

```python
import re 
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
from bs4 import BeautifulSoup
```

```python
### Step 2 -- Clean the data -- 
# Removing special characters
data['reviewText'] =data['reviewText'].apply(lambda x:re.sub('[^a-z A-Z 0-9]+','',x))
# Removing the stipwords
data['reviewText'] =data['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
# Remove url
data['reviewText'] =data['reviewText'].apply(lambda x:re.sub(r'(http|https|ftp|://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?())','',str(x)))
# Remove html tags
data['reviewText'] =data['reviewText'].apply(lambda x: BeautifulSoup(x,'lxml').get_text())
# Remove Any additional spaces
data['reviewText'] =data['reviewText'].apply(lambda x:" ".join(x.split()))
```

```python
data.head()
```

```python
### Lemmatizer
from nltk.stem import WordNetLemmatizer
lemma = WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemma.lemmatize(word) for word in text.split()])
```

```python
data['reviewText'] =data['reviewText'].apply(lambda x:lemmatize_words(x))
```

```python
data.head()
```

```python
### Train-Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(data['reviewText'],data['rating'],
                                               test_size=0.20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow = CountVectorizer()
X_train_bow = bow.fit_transform(X_train).toarray()
X_test_bow =bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
X_train_tfidf = tfidf.fit_transform(X_train).toarray()
X_test_tfidf =tfidf.transform(X_test).toarray()
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow = GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf = GaussianNB().fit(X_train_tfidf,y_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
y_pred_bow = nb_model_bow.predict(X_test_bow)
```

```python
y_pred_tfidf = nb_model_bow.predict(X_test_tfidf)
```

```python
from gensim.models import Word2Vec
import numpy as np
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
X_train_tokens = [text.split() for text in X_train]
X_test_tokens = [text.split() for text in X_test]
```

```python
w2v_model = Word2Vec(
    sentences=X_train_tokens,
    vector_size=200,
    window=10,
    min_count=1,
    workers=4
)
```

```python
def avg_word2vec(words, model, vector_size):
    vec = np.zeros(vector_size)
    count = 0
    
    for word in words:
        if word in model.wv:
            vec += model.wv[word]
            count += 1
    
    if count != 0:
        vec = vec / count
        
    return vec
```

```python
X_train_w2v = np.array([avg_word2vec(words, w2v_model, 200) for words in X_train_tokens])
X_test_w2v = np.array([avg_word2vec(words, w2v_model, 200) for words in X_test_tokens])
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train_w2v, y_train)

y_pred = model.predict(X_test_w2v)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(confusion_matrix(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100)
model.fit(X_train_w2v, y_train)

y_pred = model.predict(X_test_w2v)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
from xgboost import XGBClassifier

model = XGBClassifier()
model.fit(X_train_w2v, y_train)

y_pred = model.predict(X_test_w2v)
print("Accuracy:", accuracy_score(y_test, y_pred))
```
