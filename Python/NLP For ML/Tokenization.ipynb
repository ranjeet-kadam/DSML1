```python
import nltk
nltk.download('punkt')
corpus = """
    This is a sample text corpus for natural language processing tasks. 
    It contains multiple sentences.
    The purpose of this corpus is to demonstrate text analysis techniques."""
```

```python
nltk.download('punkt_tab')
```

```python
print(corpus)
```

```python
## Tokenization Example
from nltk.tokenize import word_tokenize, sent_tokenize
# Tokenizing into sentences
sentences = sent_tokenize(corpus, language='english')
#print(sentences)
#print(type(sentences))
for i, sent in enumerate(sentences):
    print(i+1, sent)
```

```python
## Sentence into words
documents = word_tokenize(corpus)
```

```python
for sentence in documents:
    print(sentence)
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenzier = TreebankWordTokenizer()
```

```python
tokenzier.tokenize(corpus)
```
