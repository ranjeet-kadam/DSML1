```python
### Speech of A.P.J. Abdul Kalam at IIM Shillong
paragraph ="""The topic I have selected for the interaction with you is “Creating a Liveable Planet Earth.”

The Earth is the only livable planet in our solar system. But today the Earth is facing many challenges because of the actions of human beings. Pollution, climate change, violence, and the irresponsible use of natural resources are threatening the sustainability of life on this planet.

Recently, we have also seen incidents of violence where innocent people have lost their lives. These man-made forces are as big a threat to the livability of the Earth as pollution and environmental damage. If this trend continues, humanity will face serious consequences in the future.

Therefore, the important question for all of us is:

How do we make the Earth a liveable planet for the future generations?

For that, every citizen, every student, every leader, and every nation has a responsibility. We must work together to protect the environment, reduce pollution, and promote peace.

Young people like you have a special role. Your ideas, your innovation, and your leadership can transform the world. When the youth of a nation combine knowledge with a mission, they can create a powerful force for change.

Let us work for a world where development and sustainability go together, where technology helps humanity, and where the Earth remains a safe and beautiful home for future generations."""
```

```python
import nltk
from nltk.corpus import stopwords

# Download resources
# nltk.download('punkt')
# nltk.download('stopwords')
# nltk.download('averaged_perceptron_tagger_eng')

#paragraph = """The topic I have selected for the interaction with you is “Creating a Liveable Planet Earth.”"""

sentences = nltk.sent_tokenize(paragraph)

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word.lower() not in stopwords.words('english')]
    sentences[i] = nltk.pos_tag(words)

print(sentences)
```
